```python
df = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

df["wine_type"] = wine.target
```

```python
df["wine_name"] = df["wine_type"].map(
    {
        0: "Wine A",
        1: "Wine B",
        2: "Wine C"
    }
)
```

```python
import matplotlib.pyplot as plt
```

```python
df["wine_name"].value_counts().plot(
    kind="bar"
)

plt.title("Cantidad de vinos por clase")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["alcohol"],
    df["color_intensity"]
)

plt.xlabel("Alcohol")
plt.ylabel("Color Intensity")

plt.show()
```

```python
colors = ["red", "blue", "green"]

for wine_type in df["wine_type"].unique():

    subset = df[df["wine_type"] == wine_type]

    plt.scatter(
        subset["alcohol"],
        subset["color_intensity"],
        label=f"Clase {wine_type}"
    )

plt.xlabel("Alcohol")
plt.ylabel("Color Intensity")
plt.legend()
plt.show()
```
