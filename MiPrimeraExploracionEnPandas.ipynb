```python
import pandas as pd

data = {
    'Col1': [1, 2, 3, 4, 5],
    'Col2': ['A', 'B', 'C', 'D', 'E'],
    'Col3': [10.1, 20.2, 30.3, 40.4, 50.5]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.describe())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.hist(df['Col3'], bins=5, edgecolor='black')
plt.title('Histogram of Col3')
plt.xlabel('Col3 Values')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
