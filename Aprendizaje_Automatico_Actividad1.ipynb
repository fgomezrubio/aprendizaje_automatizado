características que podrían influir en su desempeño académico.

La variable `Exam_Score` puede utilizarse como variable objetivo porque
representa el puntaje obtenido por cada estudiante en un examen.

Debido a que `Exam_Score` es una variable numérica, podría plantearse un
problema de aprendizaje supervisado de tipo regresión.

Un posible modelo permitiría estimar el puntaje de examen de un estudiante
a partir de sus características académicas, personales y escolares.

### 8) Valores extremos


````python
def detectar_outliers(df,col):

    q1=df[col].quantile(.25)
    q3=df[col].quantile(.75)
    iqr=q3-q1
    li=q1-1.5*iqr
    ls=q3+1.5*iqr

    return df[(df[col]<li)|(df[col]>ls)]
````

````python
for col in numerical_cols:

    out=detectar_outliers(df,col)
    print(col,len(out))
````

Antes de modificar cualquier dato, es una buena práctica conservar el original.

````python
# Crear una copia para trabajar durante la limpieza
df_clean = df.copy()

print("Copia del dataset creada correctamente.")
````

### Paso 2. Imputar valores faltantes

Ya vimos que las tres variables con datos faltantes son:

* `Parental_Education_Level`
* `Teacher_Quality`
* `Distance_from_Home`

Todas son **variables categóricas**.

Como el porcentaje de datos faltantes es muy pequeño (≈1 %), una estrategia sencilla y apropiada para esta actividad es **imputar con la moda**.

## ¿Qué es la moda?

La moda es el valor que más se repite.

Por ejemplo:

| Variable        | Valores                           |
| --------------- | --------------------------------- |
| Teacher_Quality | High, Medium, Medium, Low, Medium |

La moda es:

```text
Medium
```



````python
# Variables categóricas con valores faltantes
categorical_missing = [
    "Parental_Education_Level",
    "Teacher_Quality",
    "Distance_from_Home"
]

for column in categorical_missing:

    mode = df_clean[column].mode()[0]
    df_clean[column] = df_clean[column].fillna(mode)
    print(f"{column}: imputada con '{mode}'")
````

## Verificación

Siempre debemos comprobar que la limpieza funcionó.


````python
print("Valores Originales: ")
df[categorical_missing].isnull().sum()
````

````python
print("Valores después de la limpieza")
df_clean[categorical_missing].isnull().sum()
````

### Interpretación

Se identificaron valores faltantes únicamente en tres variables categóricas. Debido a que el porcentaje de ausencia fue inferior al 2 % en todos los casos, se decidió reemplazar los valores faltantes utilizando la moda de cada variable. Esta estrategia conserva el número de registros y mantiene la categoría más representativa de cada atributo.

---

### ¿Por qué no eliminar filas?

Supongamos que eliminamos todas las filas con algún dato faltante.

Perderíamos información útil.

En este caso:

* el porcentaje de datos faltantes es muy bajo;
* las variables son categóricas;
* la moda representa una solución sencilla y ampliamente utilizada en ejercicios introductorios.

Por eso **la imputación es mejor que eliminar registros**.

````python
def calcular_mediana_manual(columna):
    # 1. Eliminar valores nulos (NaN) para evitar errores en el cálculo
    columna_limpia = columna.dropna()

    # 2. Ordenar los datos de menor a mayor y reiniciar el índice
    datos_ordenados = columna_limpia.sort_values().reset_index(drop=True)

    # 3. Contar el número total de observaciones
    n = len(datos_ordenados)

    # Si el DataFrame está vacío, regresar None
    if n == 0:
        return None

    # 4. Aplicar la lógica según si la cantidad de datos es par o impar
    if n % 2 != 0:
        # Caso IMPAR: Tomamos la posición del centro exacto
        posicion_medio = n // 2
        mediana = datos_ordenados.iloc[posicion_medio]
    else:
        # Caso PAR: Tomamos las dos posiciones centrales y promediamos
        posicion_centro_izquierda = (n // 2) - 1
        posicion_centro_derecha = n // 2

        valor1 = datos_ordenados.iloc[posicion_centro_izquierda]
        valor2 = datos_ordenados.iloc[posicion_centro_derecha]

        mediana = (valor1 + valor2) / 2

    return mediana
````

````python
# 5.9 Código para detectar posibles valores atípicos

#Para completar formalmente la revisión de las dos variables numéricas:

def resumen_atipicos_iqr(dataframe, columnas):
    resultados = []

    for columna in columnas:
        q1 = dataframe[columna].quantile(0.25)
        q3 = dataframe[columna].quantile(0.75)
        iqr = q3 - q1

        limite_inferior = q1 - 1.5 * iqr
        limite_superior = q3 + 1.5 * iqr

        cantidad = (
            (dataframe[columna] < limite_inferior) |
            (dataframe[columna] > limite_superior)
        ).sum()

        resultados.append({
            "Variable": columna,
            "Q1": q1,
            "Q3": q3,
            "IQR": iqr,
            "Límite inferior": limite_inferior,
            "Límite superior": limite_superior,
            "Posibles atípicos": cantidad
        })

    return pd.DataFrame(resultados)


tabla_atipicos_seleccionados = resumen_atipicos_iqr(
    df_clean,
    ["Hours_Studied", "Attendance"]
)

display(tabla_atipicos_seleccionados)
````

## Cómo interpretar el resultado

* Si `Posibles atípicos` es cero, la regla IQR no detectó valores extremos.
* Si es mayor que cero, son observaciones que quedan fuera de los límites calculados.
* La clasificación de “atípico” no equivale automáticamente a “incorrecto”.
* Antes de eliminar un valor debe comprobarse si es imposible, producto de un error o simplemente poco frecuente.

# **6. Codificación de variables categóricas**

Los algoritmos de Machine Learning generalmente requieren variables numéricas.
Por ello, las variables categóricas deben convertirse a una representación
numérica antes de utilizarse para entrenar un modelo.

No todas las variables categóricas deben transformarse de la misma manera.

En esta sección se utilizarán:

- **One-Hot Encoding** para `School_Type`, porque sus categorías no tienen
  un orden natural.
- **Codificación ordinal** para `Motivation_Level`, porque sus categorías
  representan niveles ordenados: Low, Medium y High.


## 6.1 Antes de transformar: observar los valores originales

Conviene mostrar primero qué tenemos:


````python
print("Valores de School_Type:")
print(df_clean["School_Type"].value_counts())

print("\nValores de Motivation_Level:")
print(df_clean["Motivation_Level"].value_counts())
````


Aquí debemos comprender que ambas son categóricas, pero conceptualmente son diferentes.

`School_Type` tiene algo como:

```text
Public
Private
```

No existe una relación:

```text
Public < Private
```

ni:

```text
Public > Private
```

Por tanto, es **nominal**.

En cambio `Motivation_Level` tiene:

```text
Low
Medium
High
```

y sí podemos establecer:

```text
Low < Medium < High
```

Por eso es **ordinal**.

---


# **6.2 One-Hot Encoding para `School_Type`**

````python
from sklearn.preprocessing import OneHotEncoder

#Creamos el codificador:

school_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

#Ahora realizamos la transformación:

school_encoded = school_encoder.fit_transform(
    df_clean[["School_Type"]]
)
# vemos que se produjo

print(school_encoded[:5])

````

# 6.5 Codificación ordinal para `Motivation_Level`

Ahora tenemos un problema diferente.

Tenemos:

```text

```

y queremos conservar precisamente ese orden.

Scikit-learn tiene `OrdinalEncoder`:



````python
from sklearn.preprocessing import OrdinalEncoder


# Le especificamos explícitamente el orden correcto:

motivation_encoder = OrdinalEncoder(
    categories=[
        ["Low", "Medium", "High"]
    ]
)
````

No debemos confiar simplemente en un orden alfabético. Somos nosotros quienes establecemos el **orden conceptual de la variable**.

````python
motivation_encoded = motivation_encoder.fit_transform(
    df_clean[["Motivation_Level"]]
)


## Mostramos algunos valores:

print(motivation_encoded[:10])
````

# **7. Escalamiento de variables numéricas**

Las variables numéricas de un conjunto de datos pueden encontrarse en
escalas muy diferentes.

Por ejemplo:

- Attendance se mide aproximadamente entre 60 y 100.
- Hours_Studied puede variar entre 1 y 44.

Aunque ambas variables contienen información útil, algunos algoritmos de
Machine Learning son sensibles a la magnitud de los datos y podrían dar más
importancia a la variable con valores numéricos más grandes.

Para evitar este problema, se aplicará una técnica de escalamiento.


---

# Antes de escribir una sola línea de código...

Yo les preguntaría a los alumnos:

Supongamos que tenemos dos estudiantes.

| Hours_Studied | Attendance |
| ------------: | ---------: |
|            10 |         98 |
|            30 |         65 |

Si un algoritmo calcula distancias (como KNN):

¿Qué pesa más?

La diferencia

```text
30 − 10 = 20
```

o

```text
98 − 65 = 33
```

Naturalmente la asistencia tiene mayor rango.

Entonces el algoritmo puede darle más importancia únicamente por la escala.

Eso es precisamente lo que queremos evitar.

---

# ¿Qué escalador utilizar?

La actividad permite cualquiera de estos:

* StandardScaler
* MinMaxScaler
* RobustScaler

Yo recomiendo **StandardScaler**, porque:

* es el más utilizado;
* prepara muy bien el terreno para futuros algoritmos (SVM, Regresión, Redes Neuronales, PCA);
* permite explicar la desviación estándar, que ya vimos en la sección de estadística.

---

# ¿Qué hace StandardScaler?

La idea principal es:

> Centrar los datos alrededor de cero y hacer que tengan desviación estándar igual a uno.

Es decir:

Antes:

```text
Attendance

60
72
80
94
100
```

Después:

```text
-1.75
-0.70
0.02
1.18
1.72
```

Muchos alumnos preguntan:

> ¿Cómo puede haber asistencia negativa?

La respuesta es:

**No son porcentajes nuevos.**

Son valores **escalados**.

Representan qué tan lejos está cada observación del promedio.

---

# ¿Qué significa un valor negativo?

Aquí aprovecho para conectar con la desviación estándar.

Supongamos:

Media = 80

Entonces

| Valor original |    Valor escalado |
| -------------: | ----------------: |
|             70 |          negativo |
|             80 | aproximadamente 0 |
|             90 |          positivo |

Interpretación:

* negativo → está debajo del promedio
* cero → está muy cerca del promedio
* positivo → está arriba del promedio

### PASO 1

Primero definimos qué columnas pertenecen a cada grupo.


````python
# Variables numéricas
numeric_features = [
    "Hours_Studied",
    "Attendance"
]

# Variables nominales
nominal_features = [
    "School_Type"
]

# Variables ordinales
ordinal_features = [
    "Motivation_Level"
]
````

### PASO 2

Importamos todo.


````python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    OrdinalEncoder
)
````

### PASO 3

Creamos el transformador ordinal.


````python
ordinal_encoder = OrdinalEncoder(
    categories=[
        [
            "Low",
            "Medium",
            "High"
        ]
    ]
)
````

Muy importante.

No pusimos:

```python
OrdinalEncoder()
```

Porque el orden alfabético sería

```text
High
Low
Medium
```

que está mal.

Nosotros decidimos el orden.


### PASO 4

Creamos el ColumnTransformer.

Este objeto será el corazón del preprocesamiento.

````python
preprocessor = ColumnTransformer(

    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "nominal",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            nominal_features
        ),
        (
            "ordinal",
            ordinal_encoder,
            ordinal_features
        )
    ],

    remainder="passthrough"

)
````

Literalmente:

> "Para estas columnas aplica StandardScaler."

Después

> "Para estas otras aplica OneHot."

Después

> "Para estas aplica Ordinal."

Y

```python
remainder="passthrough"
```

significa

> Las demás columnas déjalas como están.



### PASO 5

Aplicar el preprocesador.

````python
X_prepared = preprocessor.fit_transform(df_clean)
````

Eso es todo.

Un solo comando.

---

```

```

```

```


sin que nosotros escribiéramos cuatro bloques distintos.
