## Pregunta 13

¿Podemos calcular directamente el valor promedio de mercado de los jugadores utilizando la columna `Value`?

- Observa primero cómo vienen almacenados los valores.
- ¿La columna es numérica o de texto?
- ¿Qué elementos habría que eliminar o transformar antes de calcular un promedio?


```python
df["Value"].head(10)
```

```python
df["Value"].dtype
```

```python
def convertir_valor(valor):
    valor = valor.replace("€", "")

    if "M" in valor:
        return float(valor.replace("M", "")) * 1_000_000

    elif "K" in valor:
        return float(valor.replace("K", "")) * 1_000

    else:
        return float(valor)
```

```python
df["ValueNumeric"] = df["Value"].apply(convertir_valor)
```

```python
df[["Name", "Value", "ValueNumeric"]].head(10)
```

```python
df["ValueNumeric"].mean()
```

```python
df["ValueNumeric"].mean() / 1_000_000
```

## Pregunta 14

¿Podemos calcular directamente el salario promedio de los jugadores utilizando la columna `Wage`?

- Observa cómo vienen representados los salarios.
- ¿Qué transformaciones serían necesarias antes de analizarlos numéricamente?


```python
df["Wage"].head(10)
```

```python
def convertir_salario(salario):
    salario = salario.replace("€", "")

    if "K" in salario:
        return float(salario.replace("K", "")) * 1_000

    return float(salario)
```

```python
df["WageNumeric"] = df["Wage"].apply(convertir_salario)
```

```python
df[["Name", "Wage", "WageNumeric"]].head(10)
```

```python
df["WageNumeric"].mean()
```

## Pregunta 15

¿Cómo vienen almacenadas las variables `Height` y `Weight`?

- ¿Son valores numéricos?
- ¿Qué unidades aparecen?
- ¿Qué tendríamos que hacer para poder calcular promedios, mínimos, máximos o correlaciones con estas variables?


```python
df[["Height", "Weight"]].head(10)
```

```python
df[["Height", "Weight"]].dtypes
```

```python
df["HeightNumeric"] = (
    df["Height"]
    .str.replace("cm", "")
    .astype(float)
)
```

```python
df["WeightNumeric"] = (
    df["Weight"]
    .str.replace("kg", "")
    .astype(float)
)
```

```python
df[
    ["Name", "Height", "HeightNumeric", "Weight", "WeightNumeric"]
].head(10)
```

```python
df["HeightNumeric"].describe()
```

```python
df["WeightNumeric"].describe()
```

```python
df[
    ["HeightNumeric", "WeightNumeric", "Overall"]
].corr()
```
